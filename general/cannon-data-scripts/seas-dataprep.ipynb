```python
## An unnamed column gets generated while extracting data that needs to be dropped from the DF
df1_seas.drop('Unnamed: 67', axis=1, inplace=True)
df1_seas
```

```python
## For testing purposes, copy a small portion of the DF into another one and perform string operations on this before 
## implementing them on the main DF
#
dfseas_20 = df1_seas.head(20)
dfseas_20
```

```python
# Parse alphanumeric string, JobID, by splitting the leading digits from the remaining characters
# \d is for digit & \D is for that's not digit
# import re
# text = "12345abcde99xyz"
# parts = re.split(r'(^\d+)', text)
# Or 
# could work with strings
# import string
# text = "12345abcde99xyz"
# result = text.lstrip(string.digits)
# print(result)  # Output: abcde99xyz
#
import re

#pattern = r"(\d*)(\D*)"
#
# Need to extract the digits along with underscore to account for array jobs
pattern = r"([0-9_]+)(\D*)"

# Print the unique string pattern in the DF
print(dfseas_20['JobID'].str.extract(pattern)[1].unique())

# Update the DF JobID column with the extracted digital pattern
dfseas_20['JobID'] = dfseas_20['JobID'].str.extract(pattern)[0]
dfseas_20
```

```python
# Get the job array entries 
dfseas_20[dfseas_20['JobID'].str.contains('_')]
```

```python
# --- Consolidation ---
# .first() returns the first non-NaN value per column per group.
# If ALL values in a column for a given JOBID are NaN, NaN is preserved.
consolidated_dfseas = dfseas_20.groupby('JobID', as_index=False).first()
consolidated_dfseas
```

```python
# Extract the digital portion of JobID along with entries with underscore for the main DF
#
import re

pattern = r"([0-9_]+)(\D*)"
# Get a list of unique string patterns from the DF
#
print("List of unique string patterns in the DF", df1_seas['JobID'].str.extract(pattern)[1].unique())

# Extract the digits along with underscore & update JobID column with that
#
df1_seas['JobID'] = df1_seas['JobID'].str.extract(pattern)[0]
df1_seas
```

```python
# Get the job array entries with a special pattern: _[
#
df1_seas[df1_seas['JobID'].str.extract(pattern)[1]=='[']
```

```python
df1_seas[df1_seas['JobID'].str.contains('32918373')]
```

```python
# --- Consolidation ---
# .first() returns the first non-NaN value per column per group.
# If ALL values in a column for a given JOBID are NaN, NaN is preserved.
#
consol_df1seas = df1_seas.groupby('JobID', as_index=False).first()
consol_df1seas
```

```python
# Check if array jobs got correctly captured 
#
consol_df1seas[consol_df1seas['JobID'].str.contains('_')]
```

```python
# Check if '_[' pattern is still present in the updated DF
consol_df1seas[consol_df1seas['JobID'].str.extract(pattern)[1]=='[']
```

```python
# Store the consolidated filtered raw data DF into a pickle file 
#
consol_df1seas.to_pickle(rawdata)
```
